```python
def mean_std(X):
    m,n = X.shape
    mean = np.zeros(n)
    std = np.zeros(n)

    for j in range(n):
        mean[j] = np.mean(X[:,j])
        std[j] = np.std(X[:,j])
    
    return mean,std
mean,std = mean_std(X_train)
mean,std
```

```python
def Z_Score_Normalization(X):
    '''
    X_norm[j]=(X[j] - mean) / std
    X: Features X
    X_norm: Normalized Features X
    '''
    X_norm = np.zeros(X.shape)
    m,n = X.shape

    for j in range(n):
        X_norm[:,j] = (X[:,j] - mean[j]) / std[j]

    return X_norm

X_norm = Z_Score_Normalization(X_train)
y_norm = (y_train-np.mean(y_train))/np.std(y_train)
y_mean = np.mean(y_train)
y_std = np.std(y_train)
print(X_norm)
print(y_norm)
print(y_mean)
print(y_std)
```

```python
def cost_function(X,y,w,b):
    '''
    X: Matrix of size (m,n) with m examples and n features
    y: Vector of size (m,) represents labels
    w: Vector of size (n,) represents weights of n features
    b: Scalar Parameter or bias term
    
    f_wb: Precditon function
    '''

    m = X.shape[0]
    cost = 0
    for i in range(m):
        f_wb = np.dot(X[i],w)+b
        cost += (f_wb-y[i])**2
    cost = cost/(2*m)
    return cost
        
```

```python
def compute_gradient(X,y,w,b):
    '''
    Calculates the Partial derivatives w.r.t w and b
    X: Matrix of size (m,n) with m examples and n features
    y: Vector of size (m,) represents labels
    w: Vector of size (n,) represents weights of n features
    b: Scalar Parameter or bias term
    
    f_wb: Precditon function
    '''

    m,n = X.shape

    dj_dw = np.zeros((n,))
    dj_db = 0

    for i in range(m):
        cost = np.dot(X[i],w)+b - y[i]
        for j in range(n):
            dj_dw[j] = dj_dw[j] + cost*X[i,j]
        dj_db = dj_db + cost
    dj_dw = dj_dw/m
    dj_db = dj_db/m
    return dj_dw,dj_db
```

```python
def gradient_descent(X,y,w_in,b_in,gradient_function,alpha=0.000000001,num_iter=10000):

    w = w_in
    b = b_in
    
    for i in range(num_iter):
        dj_dw,dj_db = gradient_function(X,y,w,b)
        w = w-alpha * dj_dw
        b = b-alpha * dj_db
    return w,b
```
